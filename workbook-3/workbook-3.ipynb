## Функции
### 1.2.1 Пример
Напишите функцию sum_range(start, end), которая суммирует все целые числа от значения "start" до величины "end" включительно.
Если пользователь задаст первое число больше чем второе, просто поменяйте их местами.

```python
def sum_range(start,end):
    if start > end:
        end,start=start,end
    return sum(range(start,end + 1))

print(sum_range(2,12))
print(sum_range(-4,4))
print(sum_range(3,2))
```

### 1.2.2 Пример
***Задача:*** Напишите рекурсивную функцию вычисления факториала на языке Python.

```python
def fact(num):
    if num == 0:
        return 1
    else:
        return num * fact(num-1)
print(fact(5))
```

## Расстояния между объектами класса
### 1.2.3 Пример
***Задача:*** Напишите функции в Python, которая вычисляет Евклидово расстояние между двуями массивами NumPy.

```python
import numpy as np
def euclidean_distance(v1,v2):
    return sum((x-y) ** 2 for x,y in zip(v1,v2)) ** 0.5

x = np.array([0,0,0])
y = np.array([3,3,3])
print(euclidean_distance(x,y))
```

## 1.2.4 Пример
Задача: напишите 4 функции в Python, которые рассчитывают квадрат Евклидова расстояния, взвешенное евклидово расстояние, Хеммингово расстояние и расстояние Чебышева между двумя массивами NumPy.

```python
def sqr_euclidean_distance(v1,v2):
    return sum((x-y)**2 for x,y in zip(v1,v2))

def weight_euclidean_distance(v1,v2,w):
    return sum((x-y)**2*s for x,y,s in zip(v1,v2,w)) ** 0.5

def mangattan_distance(v1,v2):
    return sum(abs(x-y) for (x,y) in zip(v1,v2))

def chebushev_distance(v1,v2):
    return max(abs(x-y) for (x,y) in zip (v1,v2))

x = np.array([0,0,0])
y = np.array([3,3,3])
w = np.array([0,0,1])

print(sqr_euclidean_distance(x,y))
print(weight_euclidean_distance(x,y,w))
print(mangattan_distance(x,y))
print(chebushev_distance(x,y))
```

### 1.2.6 Пример
***Задача:*** Рассчитать между двумя точками с использованием методов определения расстояний, представленных выше.

```python
#Расстояние Евклида
print(np.linalg.norm(x-y))
#Квадрат Евклидова расстояния
print(np.linalg.norm(x-y)**2)
#Расстояние Чебышева
print(np.linalg.norm(x-y,ord=np.inf))
#Расстояние Хемминга
print(np.linalg.norm(x-y,ord=1))
```

### 1.3.1 Задание
***Задача:*** Задайте 4 точки в трехмерном пространстве, рассчитайте между ними расстояния по описанным в примере выше метрикам. Отобразите точки в трехмерном пространтсве. 

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

A = np.array([0,0,0])
B = np.array([3,3,3])
C = np.array([1,4,2])
D = np.array([4,2,5])

euclidean_dist = np.linalg.norm(C - D)
sqr_euclidean_dist = np.linalg.norm(C-D)**2
chebyshev_dist = np.linalg.norm(A-B,ord=np.inf)
hemming_dist = np.linalg.norm(A-B,ord=1)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection= '3d')

print(f"Евклидово расстояние между C и D: {euclidean_dist:.4f}")
print(f"Квадрат Евклидова расстояния C и D: {sqr_euclidean_dist:.4f}")
print(f"Расстояние Чебышева между A и B: {chebyshev_dist:.4f}")
print(f"Расстояние Хемминга между A и B: {hemming_dist:.4f}")

ax.scatter(A[0], A[1], A[2], color="blue", s=50, label="Точка A")
ax.scatter(B[0], B[1], B[2], color="blue", s=50, label="Точка B")
ax.scatter(C[0], C[1], C[2], color="red", s=70, label="Точка C")
ax.scatter(D[0], D[1], D[2], color="red", s=70, label="Точка D")

plt.show()
```
