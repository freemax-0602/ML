## 3. Нормализация
### 3.2.1 Пример
***Задача:*** Прошкладируйте числовой признак в диапазон между двумя значениями 


````python
import numpy as np
from sklearn import preprocessing

feature = np.array([[-500.5], [-100.1], [0], [100.1], [909.9]])
minmax_scale = preprocessing.MinMaxScaler(feature_range = (0,1))

sclaed_feature = minmax_scale.fit_transform(feature)
sclaed_feature
````

### 3.2.2 Пример
***Задача:*** Преобразуйте признак, чтобы он имел среднее значение 0 и стандартное отклонение 1.

````python
x = np.array([[-1000.1], [-200.2], [500.5], [600.6], [9000.9]])
scaler = preprocessing.StandardScaler()
standardized = scaler.fit_transform(x)
standardized

print("Среднее:", round(standardized.mean()))
print("Стандартное отклонение:", standardized.std())
````

### 3.2.3 Пример
***Задача:***
Дан фрейм данных
``` 
    dfTest = pd.DataFrame({'A':[14.00,90.20,90.95,96.27,91.21],
    'B':[103.02,107.26,110.35,114.23,114.68],
    'C':['big','small','big','small','small']})
```
Необходимо масштабировать его числовые столбцы.

````python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
dfTest = pd.DataFrame({'A':[14.00,90.20,90.95,96.27,91.21],
'B':[103.02,107.26,110.35,114.23,114.68],
'C':['big','small','big','small','small']})
print(dfTest)
dfTest[['A','B']] = scaler.fit_transform(dfTest[['A','B']])
dfTest
````

### 3.3.2 Задание
Задача: Загрузить фрейм данных по ссылке https://raw.githubusercontent.com/akmand/datasets/master/iris.csv
Необходимо выполнить нормализацию первого числового признака (sepal_length_cm) с использованием минимаксного преобразования, а второго (sepal_width_cm) с задействованием z-масштабирования.


````python
url = 'https://raw.githubusercontent.com/akmand/datasets/master/iris.csv'
dataframe = pd.read_csv(url)
dataframe.head(5)
print(dataframe[['sepal_length_cm']])
minmax_scale = preprocessing.MinMaxScaler(feature_range=(0,1))
sclaed_feature = minmax_scale.fit_transform(dataframe[['sepal_length_cm']])
sclaed_feature
scaler = preprocessing.StandardScaler()
standardized = scaler.fit_transform(dataframe[['sepal_width_cm']])
standardized
print("Среднее:", round(standardized.mean(), 4))
print("Стандартное отклонение:",round(standardized.std()))
````
